## C. MULTI-MODEL SUBSETTING

```python
models = ['Datsun 710', 'Lotus Europa', 'Ferrari Dino']
columns = ['Model', 'mpg', 'cyl', 'hp', 'gear']
```

```python
selected_cars = cars.loc[cars['Model'].isin(models), cols]
```
